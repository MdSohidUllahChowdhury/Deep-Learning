```python
import gradio as gr

def greet(Name, is_morning, temp):

  checkMorning = "Good Morning" if is_morning else "Good evenig"
  checkTemp = f"{Name},{checkMorning}. it is {temp} today"
  convert = (temp - 32) * 5/9
  return checkTemp , round(convert,2)

demo = gr.Interface(
    fn=greet,
    inputs=["text", "checkbox", gr.Slider(0,100)],
    outputs=["text","number"],
)

demo.launch()
```
